## 2.1 Dasar Teori

Operator adalah simbol yang digunakan untuk melakukan suatu operasi pada nilai atau variabel (operand) sehingga menghasilkan nilai baru. Dalam Python, operator dibagi menjadi beberapa jenis berdasarkan fungsi dan penggunaannya, yaitu sebagai berikut.

a. Operator Aritmatika

Operator aritmatika digunakan untuk melakukan perhitungan matematika. Operator yang digunakan dalam Python antara lain penjumlahan (+), pengurangan (-), perkalian (), pembagian (/), pembagian bulat (//), sisa pembagian (%), dan perpangkatan (*).

b. Operator Assignment (Penugasan)

Operator assignment digunakan untuk memberikan nilai pada sebuah variabel atau mengubah nilai variabel tersebut. Operator yang paling dasar adalah tanda sama dengan (=). Selain itu, ada juga operator seperti +=, -=, =, /=, //=, %=, dan *= yang digunakan untuk melakukan perhitungan sekaligus memasukkan hasilnya kembali ke variabel.

c. Operator Bitwise

Operator bitwise digunakan untuk melakukan operasi pada bilangan dalam bentuk biner (bit). Operator ini terdiri dari AND (&), OR (|), XOR (^), NOT (~), geser kiri (<<), dan geser kanan (>>). Geser kiri digunakan untuk menggeser bit ke sebelah kiri, sedangkan geser kanan digunakan untuk menggeser bit ke sebelah kanan. Pada bilangan bulat, geser kiri berkaitan dengan perkalian dengan 2, sedangkan geser kanan berkaitan dengan pembagian dengan 2.

d. Operator Keanggotaan (Membership) dan Identitas

Operator keanggotaan digunakan untuk mengecek apakah suatu nilai terdapat di dalam data seperti list, tuple, atau string. Operator yang digunakan adalah in dan not in. Sedangkan operator identitas digunakan untuk mengecek apakah dua variabel mengarah pada objek yang sama. Operator identitas yang digunakan adalah is dan is not. Operator ini berbeda dengan (==) karena (==) digunakan untuk membandingkan nilai.

e. Operator Logika (Boolean)

Operator logika digunakan untuk menggabungkan beberapa kondisi yang menghasilkan nilai True atau False. Operator yang digunakan yaitu and, or, dan not. And akan bernilai True jika kedua kondisi bernilai True. Or bernilai True jika salah satu kondisi bernilai True. Sedangkan not digunakan untuk membalik nilai True menjadi False atau sebaliknya. Untuk mendapatkan operasi XOR, yaitu kondisi True ketika kedua nilai berbeda, dapat digunakan operator (!=) pada nilai boolean.

f. Operator Perbandingan (Relasi)

Operator perbandingan digunakan untuk membandingkan dua nilai. Hasil dari perbandingan ini berupa True atau False. Operator yang digunakan yaitu >, <, >=, <=, ==, dan !=. Python juga dapat melakukan perbandingan secara berantai, contohnya 0 < x < 100. Artinya, nilai x harus lebih besar dari 0 dan lebih kecil dari 100.

```python
angka1_1005 = int(input("Input angka-1: "))
angka2_1005 = int(input("Input angka-2: "))
#penjumlahan
hasil = angka1_1005 + angka2_1005
print("\nOperator Penjumlahan")
print("Hasil =", hasil)

#pengurangan
hasil = angka1_1005 - angka2_1005
print("\nOperator Pengurangan")
print("Hasil =", hasil)

#perkalian
hasil = angka1_1005 * angka2_1005
print("\nOperator Perkalian")
print("Hasil =", hasil)

#pembagian
if angka2_1005 != 0:
    hasil = angka1_1005 / angka2_1005
    print("\nOperator Pembagian")
    print("Hasil =", hasil)

    hasil = angka1_1005 // angka2_1005
    print("\nOperator Pembagian Bulat")
    print("Hasil =", hasil)

    hasil = angka1_1005 % angka2_1005
    print("\nOperator Sisa Bagi")
    print("Hasil =", hasil)
else:
    print("Angka kedua tidak boleh bernilai 0.")

#pangkat
hasil = angka1_1005 ** angka2_1005
print("\nOperator Pangkat")
print("Hasil =", hasil)
```

2.2.2 Program "Assignment_2611531005"

```python
angka1_1005 = int(input("Input angka-1: "))
angka2_1005 = int(input("Input angka-2: "))

print("\nNilai awal angka1_1005 =", angka1_1005)
print("Nilai angka2_1005 =", angka2_1005)

# Assignment biasa
hasil = angka1_1005
print("\nAssignment biasa (=)")
print("Hasil =", hasil)

# Assignment penambahan
hasil = angka1_1005
hasil += angka2_1005
print("\nAssignment penambahan (+=)")
print("Hasil =", hasil)

# Assignment pengurangan
hasil = angka1_1005
hasil -= angka2_1005
print("\nAssignment pengurangan (-=)")
print("Hasil =", hasil)

# Assignment perkalian
hasil = angka1_1005
hasil *= angka2_1005
print("\nAssignment perkalian (*=)")
print("Hasil =", hasil)

# Assignment pembagian, pembagian bulat, dan sisa bagi
if angka2_1005 != 0:
    hasil = angka1_1005
    hasil /= angka2_1005
    print("\nAssignmnet pembagian (/=)")
    print("Hasil =", hasil)
    #operator tambahan
    hasil = angka1_1005
    hasil //= angka2_1005
    print("\nA ssignment pembagian bulat (//=)")
    print("Hasil =", hasil)
    hasil = angka1_1005
    hasil %= angka2_1005
    print("\nAssignment sisa bagi (%=)")
    print("Hasil =", hasil)
else:
    print("\nPembagian tidak dapat dilakukan.")
    print("Angka kedua tidak boleh bernilai 0.")

# Operator tambahan: assignment perpangkatan
hasil = angka1_1005
hasil **= angka2_1005
print("\nAssignment perpangkatan (**=)")
print("Hasil =", hasil)
```

*    Bit 1 Jika belanja minimal Rp200.000, nilainya 2. Jika tidak, nilainya 0.

*    Bit 2 Jika jumlah barang minimal 3, nilainya 4. Jika tidak, nilainya 0.

*    Bit 3 Jika promo tersedia, nilainya 8. Jika tidak, nilainya 0.

10. Bitwise OR (|) Operator | digunakan untuk menggabungkan status-status yang ada. Nilai dari bit 0 sampai bit 3 digabungkan menjadi status_bit_1001. Hasilnya dapat menunjukkan status pelanggan dalam bentuk bilangan biner.

Contohnya jika semua kondisi terpenuhi: 0001 | 0010 | 0100 | 1000 = 1111

1111 dalam biner memiliki nilai desimal 15.

11. Bitwise AND (&) Operator & digunakan untuk memeriksa apakah suatu status tertentu aktif.

*    Pemeriksaan Member status_bit_1005 & 1 digunakan untuk memeriksa bit member. Jika hasilnya 1, berarti status member aktif. Hasil tersebut kemudian diubah menjadi Boolean menggunakan bool().

*    Pemeriksaan Promo status_bit_1005 & 8 digunakan untuk memeriksa apakah bit promo aktif. Jika hasilnya 8, berarti promo tersedia. Jika hasilnya 0, berarti promo tidak tersedia.

12. Bitwise XOR (^) Operator ^ disebut XOR (Exclusive OR). Digunakan untuk membandingkan dua nilai biner. Bit akan menghasilkan 1 jika kedua bit yang dibandingkan berbeda. Jika kedua bit sama, hasilnya 0. Pada program: status_bit_1005 dibandingkan dengan kode referensi 11. Angka 11 dalam biner adalah 1011. Hasil XOR menunjukkan bagian bit yang berbeda antara kode transaksi dan kode referensi.

13. Bitwise Shift Left (<<) Operator << digunakan untuk menggeser bit ke sebelah kiri. Pada program, status_bit_1005 digeser sebanyak 1 posisi ke kiri. Secara umum, pergeseran satu posisi ke kiri pada bilangan biner setara dengan mengalikan bilangan dengan 2. Contoh: 0011 → 0110 3 → 6

Operator aritmatika digunakan dalam beberapa proses perhitungan pada program.
